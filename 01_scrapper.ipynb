```python
from google_play_scraper import reviews, Sort
import pandas as pd
```

```python
APP_PACKAGE = "com.jatengprov.bapenda.newsakpole"
OUTPUT_FILE = "review_new_sakpole.csv"
```

```python
result, token = reviews(
    APP_PACKAGE,
    lang="id",
    country="id",
    sort=Sort.MOST_RELEVANT,
    count=40000
)
```

```python
review_data = pd.DataFrame(result)
print(review_data)
```

```python
review_data.to_csv(OUTPUT_FILE, index=False, encoding="utf-8-sig")
```

```python
print(f"Total review berhasil diambil: {len(review_data)}")
print(f"File berhasil disimpan sebagai: {OUTPUT_FILE}")
```
